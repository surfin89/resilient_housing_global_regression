```python
import os
import sys
import pandas as pd
import re
from google.colab import drive

# --- Configuration ---
PROJECT_ROOT = "/content/drive/MyDrive/Resilient_Housing_Global_Regression"
DATA_RAW_DIR = os.path.join(PROJECT_ROOT, "data_raw", "Global_MPI_2025")
DATA_PROCESSED_DIR = os.path.join(PROJECT_ROOT, "data_processed")
INPUT_FILENAME = "MPI_national.xlsx"
OUTPUT_FILENAME = "country_MPI.csv"

# 1. Mount Google Drive
drive.mount('/content/drive')

# 2. Check and Set Working Directory
if os.path.exists(PROJECT_ROOT):
    os.chdir(PROJECT_ROOT)
    if PROJECT_ROOT not in sys.path:
        sys.path.append(PROJECT_ROOT)
    print(f"✅ Connected to Project Root: {os.getcwd()}")
else:
    print(f"❌ Error: Project directory not found at {PROJECT_ROOT}")

# 3. Ensure Output Directory Exists
os.makedirs(DATA_PROCESSED_DIR, exist_ok=True)
```

```python
def clean_national_mpi_data(input_path, output_path):
    """
    Cleans Global MPI National data by normalizing headers,
    handling missing values, and ensuring UTF-8 encoding.

    Args:
        input_path (str): Path to the raw Excel file.
        output_path (str): Path to save the cleaned CSV file.
    """
    print(f"🚀 Starting cleaning process for: {os.path.basename(input_path)}")

    try:
        # Load the dataset
        # header=0 assumes the first row contains column names
        df = pd.read_excel(input_path, header=0)

        # --- 1. Header Normalization ---
        # Strip whitespaces, remove newlines, and collapse multiple spaces
        df.columns = [re.sub(r'\s+', ' ', str(col).strip()) for col in df.columns]

        # --- 2. Handling Missing Values ---
        # Replace NaN with 'NUL' as per project requirement
        # Note: Professional projects often use specific placeholders or handle nulls per column type
        df = df.fillna('NUL')

        # --- 3. Data Export ---
        # utf-8-sig ensures compatibility with Excel on both Windows and Mac
        df.to_csv(output_path, index=False, encoding='utf-8-sig')

        print(f"✅ Successfully exported cleaned data to: {output_path}")
        print(f"📊 Dataset Shape: {df.shape}")

    except FileNotFoundError:
        print(f"❌ Error: File not found at {input_path}. Please check the path.")
    except Exception as e:
        print(f"❌ An unexpected error occurred: {e}")

# Define paths for execution
target_input = os.path.join(DATA_RAW_DIR, INPUT_FILENAME)
target_output = os.path.join(DATA_PROCESSED_DIR, OUTPUT_FILENAME)
```

```python
# Execute the cleaning function
clean_national_mpi_data(target_input, target_output)

# Preview the results
if os.path.exists(target_output):
    preview_df = pd.read_csv(target_output)
    print("\n--- Processed Data Preview (First 5 Rows) ---")
    display(preview_df.head())
```
