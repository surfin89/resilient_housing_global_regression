```python
import pandas as pd
import os
import re # 正規表現を使うためにインポート
import numpy as np

# --- 設定変数の定義 ---
# ベースディレクトリ
BASE_DATA_DIR = r'C:\Users\wb512463\OneDrive - WBG\Resilient Housing\Global analysis'

# 入力ファイルのパス
SUB_NATIONAL_INPUT_PATH = os.path.join(BASE_DATA_DIR, r'data_raw\Global MPI 2025\MPI_subnational.xlsx')
NATIONAL_INPUT_PATH = os.path.join(BASE_DATA_DIR, r'data_raw\Global MPI 2025\MPI_national.xlsx')

# 出力先のフォルダパス
OUTPUT_DIR = os.path.join(BASE_DATA_DIR, 'data_processed')

# 出力CSVファイル名
SUB_NATIONAL_OUTPUT_FILE_NAME = 'adm1_MPI.csv'
NATIONAL_OUTPUT_FILE_NAME = 'country_MPI.csv'

# --- データ処理関数 ---

def clean_and_export_mpi_data(input_path, output_file_name, output_dir, null_placeholder='NUL'):
    """
    MPIデータを読み込み、コラム名をクリーンアップし、欠損値を置換した後、CSVとしてエクスポートする。

    Args:
        input_path (str): 入力Excelファイルの完全パス。
        output_file_name (str): 出力CSVファイル名。
        output_dir (str): 出力先フォルダのパス。
        null_placeholder (str): 欠損値 (NaN) を置換する文字列。
    """
    
    # 出力ファイルの完全パス
    output_file_path = os.path.join(output_dir, output_file_name)

    print(f"\n--- 処理開始: {output_file_name} ---")

    try:
        # 1. Excelファイルを読み込み (1行目をコラム名として指定)
        print(f"📄 Excelファイルを読み込み中: {os.path.basename(input_path)}")
        df = pd.read_excel(input_path, header=0)

        # 2. 【✨コラム名（行名）のクリーンアップ✨】
        print("🧹 コラム名（行名）のクリーンアップを開始します...")
        
        new_columns = []
        for col in df.columns:
            # a. 先頭/末尾の空白を除去 (strip)
            cleaned_col = str(col).strip()
            
            # b. 改行コード (\n, \r) やタブ (\t) を全て単一の空白に置換
            cleaned_col = re.sub(r'[\n\r\t]', ' ', cleaned_col)
            
            # c. 連続する複数の空白を単一の空白に置換
            cleaned_col = re.sub(r'\s+', ' ', cleaned_col)
            
            new_columns.append(cleaned_col)
            
        # クリーンアップされたコラム名に置き換え
        df.columns = new_columns
        print("✅ コラム名のクリーンアップが完了しました。")

        # 3. 欠損値 (NaN) を指定の文字列に置換
        # df.replace([np.nan, None], null_placeholder, inplace=True) # Noneも考慮する場合
        df = df.fillna(null_placeholder)
        print(f"✅ 欠損値 (NaN) を '{null_placeholder}' に置換しました。")

        # 4. CSVファイルとして出力
        df.to_csv(output_file_path, index=False, encoding='utf-8-sig')

        print(f"\n✅ データベースCSVファイルの再作成が完了しました。")
        print(f"出力ファイルパス: {output_file_path}")
        
        # クリーンアップ後のコラム名の一部を表示して確認
        print("💡 クリーンアップ後のコラム名（一部）:")
        print(df.columns[:5].tolist())

    except FileNotFoundError:
        print(f"\n❌ エラー: 指定された入力ファイルが見つかりませんでした。パスを確認してください: {input_path}")
    except Exception as e:
        print(f"\n❌ エラーが発生しました: {e}")

# --- メイン実行ブロック ---

if __name__ == '__main__':
    
    # 0. 出力先フォルダが存在しない場合は作成 (両方の処理の前に一度だけ行う)
    if not os.path.exists(OUTPUT_DIR):
        os.makedirs(OUTPUT_DIR)
        print(f"📁 出力フォルダを作成しました: {OUTPUT_DIR}")

    # 1. サブナショナルデータの処理とエクスポート
    clean_and_export_mpi_data(
        input_path=SUB_NATIONAL_INPUT_PATH,
        output_file_name=SUB_NATIONAL_OUTPUT_FILE_NAME,
        output_dir=OUTPUT_DIR,
        null_placeholder='NUL'
    )

    # 2. ナショナルデータの処理とエクスポート (新たに追加)
    clean_and_export_mpi_data(
        input_path=NATIONAL_INPUT_PATH,
        output_file_name=NATIONAL_OUTPUT_FILE_NAME,
        output_dir=OUTPUT_DIR,
        null_placeholder='NUL'
    )
    
    print("\n\n🎉 すべてのデータ処理が完了しました。")
    print("⚠️ 注意: データ分析を行う際は、'NUL'という文字列の欠損値が含まれることを考慮して処理してください。")
```
